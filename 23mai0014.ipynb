```python
df = pd.read_csv("mllab1.csv")

fig,(g1,g2) = plt.subplots(1,2,dpi=95)

g1.boxplot(df['Price'])
g1.set_title("Price")
g1.set_ylabel("Price")
g2.boxplot(df['Quantity'])
plt.show()
```

```python
df.isnull().sum()
```

```python
df['Quantity'].isnull()
```

```python
df['Quantity'].isnull().sum()
df['Quantity'] = df['Quantity'].fillna(0)
df['Quantity'].isnull().sum()

df['Price'].isnull().sum()
df['Price'] = df['Price'].fillna(0)
df['Price'].isnull().sum()
```

```python
df.isnull().sum()
```

```python
fig,(g1,g2) = plt.subplots(1,2,dpi=95)

g1.boxplot(df['Price'])
g1.set_title("Price")
g1.set_ylabel("Price")
g2.boxplot(df['Quantity'])
g2.set_title("Quantity")
g2.set_ylabel("Quantity")
plt.show()
```

```python
q1,q3 = np.percentile(df['Price'],[25,75])

iqr = q3 - q1

lower = q1 - (1.5*iqr)
upper = q3 + (1.5*iqr)

u_array = np.where(df['Price'] >= upper)[0]
l_array = np.where(df['Price'] <= lower)[0]

print(u_array)
print(l_array)
```

```python
df.drop(index= u_array, inplace = True)
df.drop(index= l_array, inplace = True)

fig = plt.figure()
plt.boxplot(df['Price'])
plt.show()
```

```python
df['ProductCategory']= df['ProductCategory'].astype('category')
df['PCNew'] = df['ProductCategory'].cat.codes
print(df['PCNew'])
df.head(20)
```
